## Завдання 3. Скалярний добуток і норма вектора

Для векторів `u = [3, -1, 2]` та `v = [1, 4, -2]`:

1. Обчисліть скалярний добуток.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами за формулою

\[
\cos(\alpha)=\frac{u\cdot v}{\|u\|\|v\|}
\]


````python
u = np.array([3, -1, 2])
v = np.array([1, 4, -2])

dot_product = np.dot(u, v)
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
cos_alpha = dot_product / (norm_u * norm_v)

print("Скалярний добуток u · v:", dot_product)
print("Норма ||u||:", norm_u)
print("Норма ||v||:", norm_v)
print("cos(alpha):", cos_alpha)
````

## Завдання 8. Визначник та обернена матриця

Для матриці

\[
A =
\begin{bmatrix}
4 & 7\\
2 & 6
\end{bmatrix}
\]

1. Знайдіть визначник.
2. Знайдіть обернену матрицю.
3. Перевірте результат множенням `A @ A_inv`.
4. Поясніть у коментарі, який результат має бути отриманий.

````python
A = np.array([[4, 7], [2, 6]])

det_A = np.linalg.det(A)
A_inv = np.linalg.inv(A)
product_check = A @ A_inv

# Очікується одинична матриця розміру 2x2 (з точністю до похибки обчислень з рухомою комою):
# [[1., 0.],
#  [0., 1.]]
print("Визначник det(A):", det_A)
print("Обернена матриця A_inv:\n", A_inv)
print("Перевірка A @ A_inv:\n", product_check)
````

## Завдання 12. Власна функція

Напишіть функцію `vector_stats(x)`, яка приймає числовий вектор і повертає:

- мінімум;
- максимум;
- середнє;
- медіану;
- стандартне відхилення.

Перевірте функцію на векторі:

`[4, 7, 1, 9, 3, 8, 5]`.

````python
def vector_stats(x):
    arr = np.asarray(x)
    return {
        "min": np.min(arr),
        "max": np.max(arr),
        "mean": np.mean(arr),
        "median": np.median(arr),
        "std": np.std(arr),
    }


test_vector = np.array([4, 7, 1, 9, 3, 8, 5])
stats = vector_stats(test_vector)
for k, v in stats.items():
    print(f"{k}: {v}")
````

## Завдання 13. Функція з умовою

Створіть функцію

\[
f(x) =
\begin{cases}
x^2, & x < 0 \\
2x + 1, & x \geq 0
\end{cases}
\]

Застосуйте її до вектора значень від `-5` до `5`.

Спробуйте реалізувати функцію двома способами:
1. через звичайну функцію Python;
2. через `np.where()`.

````python
x = np.arange(-5, 6)


# Варіант 1: звичайна функція Python
def f_py(val):
    if val < 0:
        return val**2
    return 2 * val + 1


res_py = np.array([f_py(val) for val in x])

# Варіант 2: через np.where()
res_where = np.where(x < 0, x**2, 2 * x + 1)

print("x:", x)
print("Варіант 1 (Python):", res_py)
print("Варіант 2 (np.where):", res_where)
````

## Завдання 15. Підсумкове

Згенеруйте матрицю `5 × 5` із випадкових цілих чисел від `1` до `100`.

Виконайте такі дії:
1. Знайдіть середнє значення кожного рядка.
2. Знайдіть середнє значення кожного стовпця.
3. Замініть усі елементи, менші за загальне середнє матриці, на `0`.
4. Знайдіть рядок із найбільшою сумою.
5. Нормалізуйте початкову матрицю за формулою:

\[
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

6. Напишіть функцію, яка виконує таку нормалізацію для довільної числової матриці.

````python
rng = np.random.default_rng(42)
matrix = rng.integers(1, 101, size=(5, 5))

print("Початкова матриця:")
print(matrix)

# 1. Середнє значення кожного рядка
row_means = matrix.mean(axis=1)

# 2. Середнє значення кожного стовпця
col_means = matrix.mean(axis=0)

# 3. Заміна елементів, менших за загальне середнє матриці, на 0
overall_mean = matrix.mean()
matrix_thresholded = np.where(matrix < overall_mean, 0, matrix)

# 4. Рядок із найбільшою сумою (0-based індекс і сам рядок)
row_sums = matrix.sum(axis=1)
max_row_idx = int(np.argmax(row_sums))
max_row = matrix[max_row_idx]

# 5. Нормалізація (функція для довільної числової матриці)
def normalize_matrix(mat):
    arr = np.asarray(mat, dtype=float)
    min_v = arr.min()
    max_v = arr.max()
    if max_v == min_v:
        return np.zeros_like(arr, dtype=float)
    return (arr - min_v) / (max_v - min_v)

matrix_norm = normalize_matrix(matrix)

print("\n1. Середнє по рядках:", row_means)
print("2. Середнє по стовпцях:", col_means)
print("3. Заміна елементів < загального середнього на 0:\n", matrix_thresholded)
print(
    f"4. Рядок з найбільшою сумою: індекс {max_row_idx} (сума {row_sums[max_row_idx]}), рядок {max_row}"
)
print("5. Нормалізована матриця:\n", matrix_norm)
````
