## 13. Лінійна алгебра

### Матричний добуток

```python
a = np.array([
    [1, 2, 3],
    [0, 1, 2]
])

b = np.array([
    [ 1, 2],
    [ 0, 4],
    [-1, 0]
])

c = a @ b
print(c)
```

### Діагональ, транспонування та визначник

```python
a = np.array([
    [ 1.,  0., 3.],
    [-1., -1., 2.],
    [ 4.,  7., 2.]
])

print("diag:", np.diag(a))
print("\nA.T =\n", a.T)

det = np.linalg.det(a)
print("\ndet(A) =", det)
print("det близький до -25:", np.isclose(det, -25.0))
```

### Розв’язання системи `Ax = b`

```python
a = np.array([
    [1.,  2.,  5.],
    [1., -1.,  3.],
    [3., -6., -1.]
])

b = np.array([-9., 2., 25.])

x = np.linalg.solve(a, b)

print("x =", x)
print("Перевірка A @ x ≈ b:", np.allclose(a @ x, b))
```

Для розв’язання системи не потрібно явно обчислювати `A⁻¹`.  
У числових обчисленнях перевага надається `np.linalg.solve(A, b)`.


### Метод найменших квадратів

```python
x = np.array([
    [1., 1.],
    [1., 2.],
    [1., 3.],
    [1., 4.]
])

y = np.array([2.1, 3.9, 6.2, 8.1])

beta, residuals, rank, s = np.linalg.lstsq(x, y, rcond=None)

print("beta:", beta)
print("residuals:", residuals)
print("rank:", rank)
print("singular values:", s)
```

### Власні значення та власні вектори

```python
a = np.array([
    [2., 1.],
    [1., 2.]
])

values, vectors = np.linalg.eigh(a)

print("Власні значення:", values)
print("Власні вектори:\n", vectors)
```

### SVD

```python
a = np.array([
    [3., 1.],
    [1., 3.],
    [1., 1.]
])

U, s, Vt = np.linalg.svd(a, full_matrices=False)

print("U =\n", U)
print("\ns =", s)
print("\nVt =\n", Vt)
```

## 18. Узагальнюючий приклад: підготовка даних і лінійний прогноз

```python
# 1. Вхідні дані: два фактори та цільова змінна
x = np.array([
    [10., 2.],
    [12., 3.],
    [15., 4.],
    [18., 5.],
    [20., 6.]
])

y = np.array([21., 25., 31., 36., 40.])

# 2. Фільтрація
mask = np.isfinite(x).all(axis=1) & np.isfinite(y)
x = x[mask]
y = y[mask]

# 3. Стандартизація ознак
mu = x.mean(axis=0)
sigma = x.std(axis=0)
x_std = (x - mu) / sigma

# 4. Додаємо стовпець одиниць для вільного члена
x_design = np.column_stack([
    np.ones(x_std.shape[0]),
    x_std
])

# 5. Оцінювання параметрів методом найменших квадратів
beta, *_ = np.linalg.lstsq(x_design, y, rcond=None)

# 6. Прогноз і MSE
y_pred = x_design @ beta
mse = np.mean((y - y_pred) ** 2)

print("beta =", beta)
print("y_pred =", y_pred)
print("MSE =", mse)
```

## 19. Підсумкові тренувальні завдання

1. Створіть матрицю `X` форми `(5, 4)` із випадкових чисел.
2. Обчисліть середнє та стандартне відхилення кожної ознаки.
3. Стандартизуйте матрицю через broadcasting.
4. Виберіть рядки, у яких значення першої ознаки додатне.
5. Додайте до матриці новий стовпець одиниць.
6. Створіть вектор ваг і обчисліть прогноз `X @ w`.
7. Для заданого `y_true` обчисліть MSE.
8. Перевірте, чи немає в масивах `NaN` або `inf`.


```python
import numpy as np

np.set_printoptions(precision=3, suppress=True)
rng = np.random.default_rng(42)

# 1. Матриця X форми (5, 4) із випадкових чисел (5 об'єктів, 4 ознаки)
X = rng.normal(loc=10.0, scale=4.0, size=(5, 4))

# 2. Середнє та стандартне відхилення кожної ознаки (вздовж осі рядків axis=0 -> форма (4,))
mean = X.mean(axis=0)
std = X.std(axis=0)

# 3. Стандартизація матриці через broadcasting: (X - mean) / std
# mean та std мають форму (4,), broadcasting вирівнює їх до (1, 4)
X_scaled = np.where(std != 0, (X - mean) / std, 0.0)

# 4. Рядки, у яких значення першої ознаки (стовпець 0) додатне
pos_first_feature_rows = X[X[:, 0] > 0]

# 5. Додавання нового стовпця одиниць до матриці X (наприклад, для зсуву / bias)
ones_col = np.ones((X.shape[0], 1), dtype=X.dtype)
X_with_bias = np.hstack([X, ones_col])  # форма стає (5, 5)

# 6. Вектор ваг w та прогноз X_with_bias @ w (форма (5, 5) @ (5,) -> (5,))
w = np.array([0.5, -1.2, 0.8, 2.0, 1.5])
y_pred = X_with_bias @ w

# 7. Обчислення MSE для заданого y_true
y_true = np.array([12.0, 8.5, 15.0, 10.2, 14.1])
mse = np.mean((y_true - y_pred) ** 2)

# 8. Перевірка на відсутність NaN або inf у масивах
arrays_to_check = {
    "X": X,
    "X_scaled": X_scaled,
    "pos_first_feature_rows": pos_first_feature_rows,
    "X_with_bias": X_with_bias,
    "w": w,
    "y_pred": y_pred,
    "y_true": y_true,
}

has_nan_or_inf = {
    name: bool(np.any(~np.isfinite(arr))) for name, arr in arrays_to_check.items()
}

print("1. Матриця X (5, 4):\n", X)
print("\n2. Середнє по ознаках:", mean)
print("   Std по ознаках:", std)
print("\n3. Стандартизована матриця X_scaled:\n", X_scaled)
print("\n4. Рядки, де перша ознака > 0:\n", pos_first_feature_rows)
print("\n5. Матриця з доданим стовпцем одиниць (5, 5):\n", X_with_bias)
print("\n6. Прогноз y_pred:", y_pred)
print("\n7. MSE:", mse)
print("\n8. Чи є NaN або inf у кожному масиві:", has_nan_or_inf)
print("   У всіх масивах немає NaN чи inf:", not any(has_nan_or_inf.values()))
```

2. Атрибути: shape — розміри по осях; ndim — кількість вимірів; size — загальна кількість елементів; dtype — тип елементів; itemsize — розмір одного елемента в байтах.

3. arange vs linspace: arange(start, stop, step) задає крок і не включає stop; linspace(start, stop, num) задає кількість точок і за замовчуванням включає обидві межі.

4. view vs copy: view ділить дані з оригіналом (зміни впливають на обидва); copy створює незалежну копію даних.

5. Векторизація: виконання операцій над цілими масивами одразу у скомпільованому коді без явних циклів for у Python.

6. Broadcasting: вирівнювання shape справа наліво; осі сумісні, якщо вони однакові або одна з них дорівнює 1 («розтягується» без копіювання).

7. axis=0 і axis=1: у 2D-масиві axis=0 — уздовж рядків (результат по стовпцях), axis=1 — уздовж стовпців (результат по рядках).

8. * vs @: * — поелементне множення; @ — матричний добуток (або скалярний для 1D-векторів).

9. np.linalg.solve для $Ax = b$: швидше, точніше (менша похибка округлення) і не рахує зайву обернену матрицю inv(A).

10. np.random.default_rng(): сучасний генератор (PCG64), краща статистика, відтворюваність без залежності від глобального стану np.random.seed.
