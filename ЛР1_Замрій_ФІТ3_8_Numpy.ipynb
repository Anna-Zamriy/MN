````python
a = np.array([1, 2])
b = np.array([2, 1])

cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle_degrees = np.degrees(np.arccos(cos_angle))

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle_degrees)
````

### Завдання 28
Для матриці:

```text

     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
A = np.array([[4, 7],
              [2, 6]])

det_A = np.linalg.det(A)
print(f"Визначник матриці A: {det_A:.4f}")

inv_A = np.linalg.inv(A)
print("\nОбернена матриця A^-1:")
print(inv_A)

check_matrix = A @ inv_A
print("\nПеревірка (A @ A^-1):")
print(np.round(check_matrix))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]], dtype=float)

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)
result = (A - mu) / sigma

print("Нормалізована матриця:")
print(result)
````

```text
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
sales = np.array([[120, 150, 170, 160],
                  [80, 90, 100, 110],
                  [200, 180, 210, 230]])

total_per_product = np.sum(sales, axis=1)
total_per_month = np.sum(sales, axis=0)
best_product_idx = np.argmax(total_per_product)
best_month_idx = np.argmax(total_per_month)
mean_per_product = np.mean(sales, axis=1)

print("Загальний обсяг продажів кожного товару:", total_per_product)
print("Загальний обсяг продажів у кожному місяці:", total_per_month)
print(f"Товар із найбільшими сумарними продажами: Товар #{best_product_idx + 1} (індекс {best_product_idx})")
print(f"Місяць із найбільшим загальним обсягом продажів: Місяць #{best_month_idx + 1} (індекс {best_month_idx})")
print("Середній обсяг продажів кожного товару:", mean_per_product)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
matrix = np.random.randint(1, 21, size=(5, 5))

print("Початкова матриця 5x5:")
print(matrix)

main_diag_sum = np.trace(matrix)
anti_diag_sum = np.sum(np.diagonal(np.fliplr(matrix)))
max_per_row = np.max(matrix, axis=1)
min_per_col = np.min(matrix, axis=0)
mean_val = np.mean(matrix)
modified_matrix = matrix.copy()
modified_matrix[modified_matrix < mean_val] = 0

print(f"Сума головної діагоналі: {main_diag_sum}")
print(f"Сума побічної діагоналі: {anti_diag_sum}")
print("Максимальний елемент кожного рядка:", max_per_row)
print("Мінімальний елемент кожного стовпця:", min_per_col)
print(f"Середнє значення всієї матриці: {mean_val:.2f}")
print("Модифікована матриця (елементи < середнього замінено на 0):")
print(modified_matrix)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
A = np.random.randint(1, 10, size=(4, 4)).astype(float)

print("Початкова матриця 4x4:")
print(A)

row_norms = np.linalg.norm(A, axis=1, keepdims=True)
print("\nНорма кожного рядка:")
print(row_norms)

normalized_matrix = A / row_norms
print("\nМатриця з нормалізованими рядками:")
print(normalized_matrix)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([[1,  2,  1],
              [2,  1, -1],
              [3, -1,  2]])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
print("Знайдений розв'язок (значення x, y, z):")
print(x)

check_b = A @ x
print("\nРезультат множення матриці A на вектор розв'язків (A @ x):")
print(check_b)

is_equal = np.allclose(check_b, b)
print(f"\nПорівняння з вектором b (чи збігаються результати?): {is_equal}")
````
